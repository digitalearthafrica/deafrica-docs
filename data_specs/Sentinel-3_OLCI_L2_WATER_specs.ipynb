# Sentinel-3 OLCI L2 Water 

**Date modified:** August 2026

## Overview

### Background

The Sentinel-3 OLCI Level-2 Water Full Resolution (OL_2_WFR) product delivers satellite-derived observations for monitoring the optical properties, biological characteristics, and quality of marine and inland waters. Acquired by the Ocean and Land Colour Instrument (OLCI) onboard the Sentinel-3A and Sentinel-3B satellites, the product provides global coverage at approximately 300-meter spatial resolution, with a revisit time of 1–2 days at the equator when both satellites are operational. OLCI’s 21 spectral bands, spanning visible to near-infrared wavelengths, are designed to support detailed observations of water colour and ocean and inland water properties.

Key outputs include water-leaving reflectance across multiple spectral bands, algal pigment (chlorophyll) concentration, Total Suspended Matter (TSM), and the diffuse attenuation coefficient (KD490), which provides an indication of water transparency. The product also includes atmospheric parameters such as aerosol optical depth, aerosol Ångström exponent, and Integrated Water Vapour (IWV), together with per-pixel quality and science flags and associated uncertainty estimates. These observations support applications such as water quality monitoring, detection and assessment of algal blooms, turbidity and sediment monitoring, aquatic ecosystem assessment, and the monitoring of changes in coastal and inland water bodies.

Comprehensive documentation—including the [Sentinel-3 Ocean Colour Level-2 Data Guide](https://user.eumetsat.int/resources/user-guides/sentinel-3-ocean-colour-level-2-data-guide), [algorithm theoretical basis documents](https://user.eumetsat.int/data/satellites/sentinel-3/ocean-colour-service/resources), [product notices and technical documentation](https://user.eumetsat.int/resources/user-guides/sentinel-3-olci-ocean-colour-processing-baseline)—is freely available to support researchers, environmental agencies, and operational monitoring services.

Digital Earth Africa (DE Africa) provides open and free access to Sentinel-3 OLCI L2 Water Full Resolution products for the African continent via its cloud-hosted infrastructure and the DE Africa Sandbox. This enables direct analysis in cloud environments, and integration with other DE Africa datasets.

### Sentinel-3 OLCI Water Measurements – Descriptions

| Name | Description |
|---|---|
| **A865** (Aerosol Ångström Exponent) | Aerosol parameter describing the size distribution of atmospheric particles. |
| **ADG443_NN** (Coloured Dissolved and Detrital Matter Absorption Coefficient) | Absorption coefficient of coloured dissolved organic matter (CDOM) and detrital material at 443 nm, estimated using a neural network algorithm. |
| **B01** (Oa01) | Sensitive to aerosol correction and improved water constituent retrieval. |
| **B02** (Oa02) | Sensitive to yellow substances, detrital pigments, and turbidity. |
| **B03** (Oa03) | Corresponds to the chlorophyll absorption maximum and supports biogeochemistry and vegetation studies. |
| **B04** (Oa04) | Sensitive to chlorophyll. |
| **B05** (Oa05) | Sensitive to chlorophyll, sediment, turbidity, and red tide. |
| **B06** (Oa06) | Chlorophyll reference (minimum) band. |
| **B07** (Oa07) | Sensitive to sediment loading. |
| **B08** (Oa08) | Corresponds to the second chlorophyll absorption maximum and is sensitive to sediment, yellow substances, and vegetation. |
| **B09** (Oa09) | Optimized for improved chlorophyll fluorescence retrieval. |
| **B10** (Oa10) | Corresponds to the chlorophyll fluorescence peak and red edge. |
| **B11** (Oa11) | Represents the chlorophyll fluorescence baseline and red-edge transition. |
| **B12** (Oa12) | Sensitive to oxygen absorption, clouds, and vegetation. |
| **B16** (Oa16) | Used for atmospheric and aerosol correction. |
| **B17** (Oa17) | Supports atmospheric and aerosol correction, cloud detection, and pixel co-registration. |
| **B18** (Oa18) | Water vapour absorption reference band and common reference band with SLSTR for vegetation monitoring. |
| **B21** (Oa21) | Sensitive to water vapour absorption and atmospheric/aerosol correction. |
| **CHL_NN** (Algal Pigment – Complex Waters) | Chlorophyll-*a* concentration estimated using a neural network algorithm for optically complex coastal and inland waters. |
| **CHL_OC4ME** (Algal Pigment – Open Waters) | Chlorophyll-*a* concentration estimated using the OC4ME algorithm for open ocean waters. |
| **IWV_W** (Integrated Water Vapour) | Column-integrated atmospheric water vapour above each observation. |
| **KD490_M07** (Diffuse Attenuation Coefficient at 490 nm) | Indicates how quickly light at 490 nm is attenuated within the water column, providing a measure of water clarity. |
| **PAR** (Photosynthetically Active Radiation) | Amount of sunlight available for photosynthesis reaching the Earth's surface. |
| **T865** (Aerosol Optical Thickness) | Measure of the amount of aerosols in the atmosphere above each observation. |
| **TSM_NN** (Total Suspended Matter) | Concentration of suspended particles in the water estimated using a neural network algorithm. |
| **dataMask** (Data Mask) | Binary mask indicating valid and invalid observations. |
