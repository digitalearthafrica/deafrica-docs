# Sentinel-3 Synergy Level-2 Vegetation

**Date modified:** November 2025

## Overview

### Background

The [Sentinel-3 SYN Level-2 VG1](https://sentiwiki.copernicus.eu/web/synergy-products) product is a 1 km VEGETATION-like daily synthesis product generated from the Sentinel-3 Synergy processing chain. It combines observations from the Ocean and Land Colour Instrument (OLCI) and the Sea and Land Surface Temperature Radiometer (SLSTR) onboard Sentinel-3A and Sentinel-3B to provide harmonised land surface observations for vegetation monitoring. The product provides Top-of-Canopy (TOC) surface reflectance and NDVI on a regular latitude–longitude grid with an approximate 1 km spatial resolution at the equator.

The [SY_2_VG1](https://documentation.dataspace.copernicus.eu/APIs/SentinelHub/Data/S3SYNL2.html#sentinel-3-syn-sy_2_vg1) product is designed as a SPOT/VEGETATION continuity product supporting long-term monitoring of vegetation dynamics across large areas. It provides daily synthesis observations using the best available pixel selection approach, based on the maximum NDVI compositing method, to improve consistency in vegetation monitoring applications. The product includes four main surface reflectance channels: blue (B0), red (B2), near-infrared (B3), and mid-infrared/SWIR (MIR). NDVI is also provided and is derived from the B3 near-infrared and B2 red channels.

In addition to reflectance and NDVI, SY_2_VG1 includes supporting atmospheric and geometric information such as aerosol optical thickness, total column ozone, total column water vapour and sun-viewing geometry. These variables are useful for interpreting atmospheric effects, filtering poor-quality observations and improving the reliability of vegetation time-series analysis.

More information on the product can be found in the Sentinel-3 Synergy Products documentation and the [Sentinel-3 SYN Land User Handbook](https://dataspace.copernicus.eu/data-collections/copernicus-sentinel-missions/sentinel-3).

### Sentinel-3 SYN Level-2 VG1 Measurements – Descriptions

Digital Earth Africa’s `s3_syn_2_vg1` product provides 13 measurements for each daily composite.

- **B0_VG1 (Blue Surface Reflectance – 450 nm)**  
  Useful for distinguishing vegetation from soil and other background surfaces.

- **B2_VG1 (Red Surface Reflectance – 645 nm)**  
  Captures the red wavelengths absorbed by chlorophyll. It is paired with near-infrared reflectance to calculate NDVI.

- **B3_VG1 (Near-Infrared Surface Reflectance – 835 nm)**  
  Sensitive to vegetation canopy structure and used with red reflectance to assess vegetation greenness.

- **MIR_VG1 (Shortwave Infrared Surface Reflectance – 1,665 nm)**  
  Sensitive to vegetation and soil moisture, supporting drought and vegetation stress monitoring.

- **NDVI_VG1 (Surface Normalised Difference Vegetation Index)**  
  Calculated from atmospherically corrected red and near-infrared reflectance. It is the primary VG1 layer for monitoring vegetation greenness and condition.

- **TOA_NDVI_VG1 (Top-of-Atmosphere NDVI)**  
  Calculated from top-of-atmosphere reflectance and used to select observations for the daily composite.

- **AG_VG1 (Aerosol Optical Thickness – 550 nm)**  
  Indicates how much airborne particles such as dust and smoke attenuate light through the atmosphere.

- **OG_VG1 (Total Column Ozone)**  
  Measures the amount of ozone through the atmospheric column above each pixel, expressed in atm·cm.

- **WVG_VG1 (Total Column Water Vapour)**  
  Measures the amount of water vapour through the atmospheric column above each pixel, expressed in g/cm².

- **SZA_VG1 (Solar Zenith Angle)**  
  Angle between the Sun and the local vertical, in degrees.

- **SAA_VG1 (Solar Azimuth Angle)**  
  Compass direction of the Sun at the time of observation, in degrees.

- **VZA_VG1 (Viewing Zenith Angle)**  
  Angle between the satellite’s viewing direction and the local vertical, in degrees.

- **VAA_VG1 (Viewing Azimuth Angle)**  
  Compass direction from the pixel toward the satellite at the time of observation, in degrees.
